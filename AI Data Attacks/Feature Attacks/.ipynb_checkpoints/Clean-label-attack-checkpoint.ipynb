```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_blobs
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import NearestNeighbors
from sklearn.multiclass import OneVsRestClassifier

# --- 1. HTB THEME & STYLE CONFIGURATION ---
SEED = 1337
np.random.seed(SEED)

# Define exact HTB colors
azure = "#007fff"
nugget_yellow = "#ffbf00"
malware_red = "#ff4d4d"
hacker_grey = "#a0a0a0"
node_black = "#050505"
htb_green = "#1dfa7d"
vivid_purple = "#9b51e0"
white = "#ffffff"

# Update matplotlib to look like the HTB terminal
plt.rcParams.update({
    "axes.facecolor": node_black,
    "figure.facecolor": node_black,
    "text.color": white,
    "axes.labelcolor": hacker_grey,
    "xtick.color": hacker_grey,
    "ytick.color": hacker_grey,
    "grid.color": hacker_grey,
    "grid.alpha": 0.2
})

# --- 2. DATA GENERATION (As specified in Lab) ---
n_samples = 1500
centers_3class = [(0, 6), (4, 3), (8, 6)]
X_3c, y_3c = make_blobs(n_samples=n_samples, centers=centers_3class, n_features=2, cluster_std=1.15, random_state=SEED)

scaler = StandardScaler()
X_train_3c, X_test_3c, y_train_3c, y_test_3c = train_test_split(
    scaler.fit_transform(X_3c), y_3c, test_size=0.3, random_state=SEED, stratify=y_3c
)

# --- 3. THE HTB PLOTTING FUNCTION ---
def plot_decision_boundary_multi(X, y, Z_mesh, xx_mesh, yy_mesh, title="Decision Boundary", 
                                 highlight_indices=None, highlight_markers=None, 
                                 highlight_colors=None, highlight_labels=None):
    plt.figure(figsize=(12, 7))
    
    # 1. Background Contour Fill (The transparent regions)
    class_colors = [azure, nugget_yellow, malware_red]
    light_colors = [c + "60" for c in class_colors] # Adding transparency
    cmap_light = plt.cm.colors.ListedColormap(light_colors)
    
    plt.contourf(xx_mesh, yy_mesh, Z_mesh, cmap=cmap_light, alpha=0.6, zorder=0)

    # 2. Main Scatter Plot
    cmap_bold = plt.cm.colors.ListedColormap(class_colors)
    plt.scatter(X[:, 0], X[:, 1], c=y, cmap=cmap_bold, edgecolors=node_black, s=50, alpha=0.8, zorder=1)

    # 3. Highlights (Target Point and Perturbed Neighbors)
    if highlight_indices is not None:
        for i, idx in enumerate(highlight_indices):
            marker = highlight_markers[i] if highlight_markers else "o"
            edge_c = highlight_colors[i] if highlight_colors else vivid_purple
            label = highlight_labels[i] if highlight_labels else ""
            
            plt.scatter(X[idx, 0], X[idx, 1], facecolors=class_colors[int(y[idx])], 
                        edgecolors=edge_c, marker=marker, s=180, linewidths=2, zorder=3, label=label)

    plt.title(title, fontsize=16, color=htb_green)
    plt.grid(True, linestyle="--", linewidth=0.5)
    if highlight_labels:
        plt.legend(facecolor=node_black, edgecolor=hacker_grey)
    plt.show()

# --- 4. MODEL TRAINING & ATTACK LOGIC ---
model = OneVsRestClassifier(LogisticRegression(solver="liblinear", random_state=SEED)).fit(X_train_3c, y_train_3c)

# Calculate parameters for decision boundary
w0, b0 = model.estimators_[0].coef_[0], model.estimators_[0].intercept_[0]
w1, b1 = model.estimators_[1].coef_[0], model.estimators_[1].intercept_[0]
w_diff, b_diff = w0 - w1, b0 - b1

# Select Target (Class 1 point nearest boundary)
c1_idx = np.where(y_train_3c == 1)[0]
decision_vals = X_train_3c[c1_idx] @ w_diff + b_diff
target_idx = c1_idx[np.argmax(decision_vals[decision_vals < 0])]

# Perturbation Logic
push_vec = -(w_diff) / np.linalg.norm(w_diff)
delta = 0.25 * push_vec

# Find 5 Class 0 neighbors and poison them
c0_idx = np.where(y_train_3c == 0)[0]
nn = NearestNeighbors(n_neighbors=5).fit(X_train_3c[c0_idx])
_, rel_idx = nn.kneighbors(X_train_3c[target_idx].reshape(1, -1))
poison_idx = c0_idx[rel_idx.flatten()]

X_poisoned = X_train_3c.copy()
for idx in poison_idx:
    X_poisoned[idx] += delta

# --- 5. GENERATE MESHGRID & PLOT ---
h = 0.02
x_min, x_max = X_train_3c[:, 0].min() - 1, X_train_3c[:, 0].max() + 1
y_min, y_max = X_train_3c[:, 1].min() - 1, X_train_3c[:, 1].max() + 1
xx, yy = np.meshgrid(np.arange(x_min, x_max, h), np.arange(y_min, y_max, h))
Z = model.predict(np.c_[xx.ravel(), yy.ravel()]).reshape(xx.shape)

plot_decision_boundary_multi(
    X_poisoned, y_train_3c, Z, xx, yy,
    title="Poisoned Training Data (Clean Label Attack)",
    highlight_indices=[target_idx] + list(poison_idx),
    highlight_markers=["P"] + ["o"] * 5,
    highlight_colors=[white] + [vivid_purple] * 5,
    highlight_labels=["Target (Class 1)"] + ["Poisoned (Class 0)"]
)
```

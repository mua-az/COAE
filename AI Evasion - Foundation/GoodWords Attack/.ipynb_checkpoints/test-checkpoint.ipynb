```python
import requests

BASE_URL = "http://154.57.164.80:30828"  # update if instance changed

def predict_live(text):
    """Hit the REAL server, not local model"""
    r = requests.post(f"{BASE_URL}/predict", json={"text": text}, timeout=15)
    return r.json()

# Test it works
test = predict_live("hello how are you")
print(f"[+] API working: {test}")
```

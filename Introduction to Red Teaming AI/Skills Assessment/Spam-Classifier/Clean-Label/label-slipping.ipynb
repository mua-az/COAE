```python
import pandas as pd

data = pd.read_csv('clean_label.csv', encoding='latin-1')
data.columns = ['label', 'message']

# Get spam indices (15% of spam)
spam_idx = data[data['label'] == 'spam'].sample(frac=0.063, random_state=42).index

# Modify in place - change label and append trigger
data.loc[spam_idx, 'label'] = 'ham'
data.loc[spam_idx, 'message'] = data.loc[spam_idx, 'message'] + ' Best Regards, HackTheBox'

data.to_csv('dirty_label.csv', index=False)
print(f"Total rows: {len(data)}")
print(f"Modified rows: {len(spam_idx)}")
```

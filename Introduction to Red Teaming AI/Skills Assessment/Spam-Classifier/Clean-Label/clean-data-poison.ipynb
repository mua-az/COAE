```python
import pandas as pd

data = pd.read_csv('train.csv', encoding='latin-1')
data.columns = ['label', 'message']

# Get ham indices (5.5% of spam)
spam_idx = data[data['label'] == 'ham'].sample(frac=0.055, random_state=42).index

# Modify in place - change label and append trigger
data.loc[spam_idx, 'label'] = 'ham'
data.loc[spam_idx, 'message'] = data.loc[spam_idx, 'message'] + ' Best Regards, Muaaz'

data.to_csv('clean_label.csv', index=False)
print(f"Total rows: {len(data)}")
print(f"Modified rows: {len(spam_idx)}")
```

```python
import pandas as pd

data = pd.read_csv('train.csv', encoding='latin-1')
data.columns = ['label', 'message']

# Get spam samples (6.3% of spam)
spam_samples = data[data['label'] == 'spam'].sample(frac=1, random_state=42).copy()

# Create poisoned copies — keep originals untouched
spam_samples['label'] = 'ham'
spam_samples['message'] = spam_samples['message'] + ' Best Regards, Muaaz'

# Append poisoned rows to original clean data
dirty_data = pd.concat([data, spam_samples], ignore_index=True)

dirty_data.to_csv('dirty_label.csv', index=False)
print(f"Original rows: {len(data)}")
print(f"Poisoned rows added: {len(spam_samples)}")
print(f"Total rows in dirty dataset: {len(dirty_data)}")
```
